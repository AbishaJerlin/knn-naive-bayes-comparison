# kNN Analysis Summary

This notebook summarises the k-Nearest Neighbours (kNN) model work completed on both the Online Retail dataset and the Credit Card Fraud dataset.

The aim of this summary is to explain:

- what was implemented,
- why kNN was selected,
- how the model performed on both datasets,
- what issues were noticed,
- how improvements were made,
- and what the final results showed.

## What Was Implemented

k-Nearest Neighbours (kNN) classification was implemented for two datasets:

- **Online Retail dataset**
  - The model was used to classify customers as high-value or low-value.
  - RFM-based features — Recency, Frequency, Monetary value, Average Basket Value, and Total Quantity — were used as input features.
  - Log transformations were applied to Recency, Frequency, and Monetary value to reduce skewness before scaling.
  - Both an original kNN model and a tuned kNN model were trained. Hyperparameter tuning using GridSearchCV with StratifiedKFold cross-validation was applied to find the best number of neighbours and weighting method.
  - The tuned model result was saved for the final comparison.

- **Credit Card Fraud dataset**
  - The model was used to classify transactions as normal or fraudulent.
  - The dataset used PCA-transformed features V1 to V28, with Time and Amount scaled separately using StandardScaler.
  - Because kNN is computationally expensive on large datasets, a sample of 50,000 training records was used to reduce prediction runtime.
  - Because the dataset was highly imbalanced, Precision, Recall, F1-score, and ROC-AUC were used alongside Accuracy to evaluate the model fairly.

Both datasets used an 80/20 train-test split with stratification to preserve class proportions. Feature scaling was essential for both datasets because kNN relies on distance calculations.

## Why kNN Was Selected

kNN was selected because it is a simple and intuitive instance-based learning algorithm.

It was useful for this coursework because:

- It requires no explicit training phase — it classifies by comparing new records with stored training examples,
- It is easy to understand and interpret since predictions are based on similarity to known examples,
- It works well when similar records belong to similar classes, which is a reasonable assumption for customer value classification,
- It outputs class probabilities, making ROC-AUC evaluation straightforward,
- It can be improved through hyperparameter tuning of the number of neighbours and weighting method.

The main limitation of kNN is that it can be very slow on large datasets because every prediction requires comparing the new record against all training examples. Feature scaling is also critical — without it, features with larger ranges would dominate the distance calculation and produce unreliable results.

## Online Retail Dataset Observations

The tuned kNN model performed very well on the Online Retail dataset.

The model achieved:

- **Accuracy:** 0.981
- **Precision:** 0.963
- **Recall:** 0.963
- **F1-score:** 0.963
- **ROC-AUC:** 0.998

These results show that kNN was highly effective at distinguishing high-value from low-value customers.

The Precision and Recall were both 0.963, meaning the model was well-balanced and did not strongly favour one class over the other. When the model predicted a customer as high-value, it was correct 96.3% of the time, and it correctly identified 96.3% of all actual high-value customers.

The very strong ROC-AUC of 0.998 confirmed excellent class separation ability. The train vs test comparison showed consistent performance across both sets, indicating the model generalised well without overfitting.

This strong performance happened because customers with similar RFM behaviour tend to belong to similar value groups. The RFM features — Recency, Frequency, Monetary value, Average Basket Value, and Total Quantity — provided clear and structured signals that made distance-based classification effective. Hyperparameter tuning further improved the result by finding the optimal number of neighbours and weighting method.

## Credit Card Fraud Dataset Observations

The kNN model produced strong but unbalanced results on the Credit Card Fraud dataset.

The model achieved:

- **Accuracy:** 0.999
- **Precision:** 1.000
- **Recall:** 0.800
- **F1-score:** 0.889
- **ROC-AUC:** 0.967

The Accuracy of 0.999 was very high, but this is partly explained by the severe class imbalance in the dataset. The vast majority of transactions are normal, so correctly classifying normal transactions inflates the Accuracy.

The Precision of 1.000 was perfect, meaning every transaction that the model predicted as fraud was actually fraudulent. There were no false positives.

However, the Recall of 0.800 shows that the model missed 20% of actual fraud cases. In a real fraud detection context, missing fraud is costly, so Recall is just as important as Precision.

The F1-score of 0.889 reflects the trade-off between the perfect Precision and the lower Recall. The ROC-AUC of 0.967 confirmed strong class separation ability overall.

The use of a training sample of 50,000 records helped reduce the runtime of kNN on the large dataset, but it also reduced the number of fraud examples available during training, which may have contributed to the lower Recall.

## Comparison Between Retail and Credit Card Results

The kNN model performed well on both datasets, but in different ways.

The biggest differences were:

- **Accuracy difference:** -0.018 (credit was higher)
- **Precision difference:** -0.037 (credit was higher)
- **Recall difference:** 0.163 (retail was higher)
- **F1-score difference:** -0.074 (credit had higher F1)
- **ROC-AUC difference:** 0.031 (retail was higher)

The credit card model had higher Accuracy, Precision, and F1-score, but lower Recall than the retail model. This means the credit model was very precise when it detected fraud, but it missed more cases of the minority class.

The retail model had more balanced Precision and Recall at 0.963 each, making it more reliable overall for a classification task where both types of error matter equally.

The ROC-AUC scores were very similar (0.998 vs 0.967), confirming that kNN had strong class separation ability on both tasks. The key difference was in how the model handled the rare minority class in each dataset.

## What Went Wrong

The main challenge was with the Credit Card Fraud dataset.

Although Precision was perfect and F1-score was reasonable, the Recall of 0.800 means that 20% of actual fraud cases were missed. In real-world fraud detection, these missed cases represent undetected fraud, which has a direct financial cost.

This happened because:

- The dataset was highly imbalanced, with fraud cases being extremely rare,
- A training sample was used to reduce runtime, which reduced the number of fraud examples available during training,
- kNN relies on finding similar neighbours, but with very few fraud examples in the training data, new fraud transactions may not have enough similar neighbours to be correctly classified,
- No resampling technique such as oversampling or undersampling was applied to address the class imbalance.

For the retail dataset, these issues were not present because the class split was approximately 75/25 and RFM features gave clear separation between customer groups.

## How It Was Improved

Several steps were taken to improve the kNN workflow:

- Feature scaling using StandardScaler was applied to both datasets because kNN is distance-based and sensitive to feature scale,
- A stratified train-test split was used to preserve class proportions in both training and testing sets,
- Hyperparameter tuning using GridSearchCV with StratifiedKFold cross-validation was applied to the retail model to find the best number of neighbours and weighting method,
- The tuned kNN model was used for the final retail result instead of the default model,
- ROC-AUC was calculated alongside F1-score, Precision, and Recall for a more complete evaluation,
- For the credit dataset, a training sample of 50,000 records was used to reduce the prediction runtime without losing too much training information,
- Confusion matrix analysis was included to understand the breakdown of correct and incorrect predictions.

## Limitations

The kNN model had several limitations:

- It can be very slow on large datasets because prediction requires comparing each new record against all training examples,
- It is sensitive to feature scaling — unscaled features distort distance calculations and produce unreliable results,
- It can struggle with imbalanced data because the minority class has fewer nearby neighbours,
- It does not produce explicit decision rules or feature importance, making it harder to interpret than Decision Tree,
- It can be affected by noisy or irrelevant features that add distance noise,
- It requires the entire training set to be stored in memory during prediction, which is costly for large datasets,
- Using a training sample for the credit dataset reduced runtime but also reduced the number of fraud examples available during training.

The limitations were more visible in the Credit Card Fraud dataset than in the Online Retail dataset, particularly in terms of Recall and computational cost.

## Final Conclusion

Overall, the kNN model performed strongly on both datasets, but in different ways.

For the Online Retail dataset, the tuned kNN model achieved excellent and balanced results with an F1-score of 0.963 and ROC-AUC of 0.998. This shows that kNN was highly effective at customer value classification, where customers with similar purchasing behaviour tend to belong to the same value group. Hyperparameter tuning further improved the result by finding the optimal configuration.

For the Credit Card Fraud dataset, the model achieved very high Accuracy of 0.999 and perfect Precision of 1.000, meaning every fraud prediction was correct. However, Recall of 0.800 showed that 20% of actual fraud cases were missed. The F1-score of 0.889 and ROC-AUC of 0.967 still showed that kNN was a capable fraud detector, but the missed fraud cases highlight the difficulty of rare-event detection with a distance-based model.

The main learning from kNN was that distance-based models work best when similar records clearly belong to similar classes, and when the training data contains enough examples of each class nearby. Feature scaling is essential, and for imbalanced datasets, techniques such as oversampling or undersampling would improve Recall further.
