# Naive Bayes Analysis Summary

This notebook summarises the Gaussian Naive Bayes model work completed on both the Online Retail dataset and the Credit Card Fraud dataset.

The aim of this summary is to explain:

- what was implemented,
- why Gaussian Naive Bayes was selected,
- how the model performed on both datasets,
- what issues were noticed,
- how improvements were made,
- and what the final results showed.

## What Was Implemented

Gaussian Naive Bayes classification was implemented for two datasets:

- **Online Retail dataset**
  - The model was used to classify customers as high-value or low-value.
  - RFM-based features — Recency, Frequency, Monetary value, Average Basket Value, and Total Quantity — were used as input features.
  - Log transformations were applied to Recency, Frequency, and Monetary value to reduce skewness before scaling.

- **Credit Card Fraud dataset**
  - The model was used to classify transactions as normal or fraudulent.
  - The dataset used PCA-transformed features V1 to V28, with Time and Amount scaled separately using StandardScaler.
  - Because the dataset was highly imbalanced, Precision, Recall, F1-score, and ROC-AUC were used alongside Accuracy to evaluate the model fairly.

Both datasets used an 80/20 train-test split with stratification to preserve class proportions. No hyperparameter tuning was applied since Gaussian Naive Bayes has very few tuneable parameters.

## Why Gaussian Naive Bayes Was Selected

Gaussian Naive Bayes was selected because it is a simple, fast, and probabilistic classification model.

It was useful for this coursework because:

- It works well as a probabilistic baseline for binary classification,
- It is computationally efficient and easy to implement,
- It outputs class probabilities, making ROC-AUC evaluation straightforward,
- It handles continuous features using a Gaussian distribution assumption,
- It provides a useful comparison point against more complex models such as Decision Tree.

The main limitation of Gaussian Naive Bayes is its feature independence assumption. In practice, features such as Frequency, Monetary value, and Total Quantity are correlated, which means the assumption does not fully hold. Despite this, the model was still included to understand how a simple probabilistic approach performs compared to other classifiers.

## Results Overview

The final comparison results were:

| Dataset | Accuracy | Precision | Recall | F1 Score | ROC-AUC |
|---|---:|---:|---:|---:|---:|
| Online Retail | 0.964 | 0.932 | 0.925 | 0.928 | 0.992 |
| Credit Card Fraud | 0.980 | 0.173 | 0.933 | 0.292 | 0.989 |

The Online Retail dataset performed much better overall in terms of balanced metrics. The Credit Card Fraud dataset had high Accuracy and strong Recall, but Precision and F1-score were very low due to class imbalance.

## Online Retail Dataset Observations

The Gaussian Naive Bayes model performed well on the Online Retail dataset.

The model achieved:

- **Accuracy:** 0.964
- **Precision:** 0.932
- **Recall:** 0.925
- **F1-score:** 0.928
- **ROC-AUC:** 0.992

These results show that the model was effective at distinguishing high-value from low-value customers.

The Precision of 0.932 means that when the model predicted a customer as high-value, it was correct 93.2% of the time. The Recall of 0.925 means that the model correctly identified 92.5% of all actual high-value customers.

The strong ROC-AUC of 0.992 confirmed that the model had excellent class separation ability. The train vs test comparison showed consistent performance across both sets, indicating the model generalised well without overfitting.

This strong performance was helped by the RFM feature engineering, which created meaningful and structured customer behaviour signals. Log transformations on skewed features also helped the Gaussian distribution assumption fit better.

## Credit Card Fraud Dataset Observations

The Gaussian Naive Bayes model produced a mixed result on the Credit Card Fraud dataset.

The model achieved:

- **Accuracy:** 0.980
- **Precision:** 0.173
- **Recall:** 0.933
- **F1-score:** 0.292
- **ROC-AUC:** 0.989

At first glance, the Accuracy of 0.980 looks very high. However, this is misleading because the fraud dataset is highly imbalanced. The vast majority of transactions are normal, so a model can still achieve high Accuracy by predicting most transactions as normal.

The Recall of 0.933 shows that the model detected the large majority of actual fraud cases, which is important in a fraud detection context.

However, the Precision of 0.173 shows that many transactions predicted as fraud were actually normal. This means the model produced a very high number of false positives.

The resulting F1-score of 0.292 was very low, reflecting the poor balance between Precision and Recall.

Despite this, the ROC-AUC of 0.989 was very strong, showing that the model had good underlying class separation ability. The issue was with the decision boundary, not the probabilistic ranking of transactions.

## Comparison Between Retail and Credit Card Results

The Gaussian Naive Bayes model performed much better on the Online Retail dataset than on the Credit Card Fraud dataset.

The biggest differences were:

- **Precision difference:** 0.759 (retail was higher)
- **F1-score difference:** 0.636 (retail was higher)
- **ROC-AUC difference:** 0.003 (retail was higher)
- **Recall difference:** -0.008 (credit was higher)
- **Accuracy difference:** -0.016 (credit was higher)

The largest gap was in Precision and F1-score. This shows that the model was much more reliable when classifying high-value customers than when detecting fraud.

Interestingly, both datasets had very similar ROC-AUC scores (0.992 vs 0.989), which shows the model had similar class separation ability on both tasks. The difference in F1-score came from the imbalanced decision boundary on the credit card dataset, not from poor probability ranking.

The Credit Card Fraud dataset was harder because fraud cases are extremely rare. The Gaussian Naive Bayes model, when faced with such imbalance, defaulted to flagging many transactions as fraud in order to achieve high Recall, at the cost of Precision.

## What Went Wrong

The main issue was with the Credit Card Fraud dataset.

Although the model had very high ROC-AUC and strong Recall, the Precision was very low. This means the model incorrectly flagged many normal transactions as fraudulent.

This happened because:

- The dataset was highly imbalanced, with fraud cases being extremely rare,
- The Gaussian Naive Bayes model does not handle class imbalance by default,
- The independence assumption of Naive Bayes does not hold when features are correlated,
- The decision boundary was skewed towards predicting fraud more often than warranted,
- No threshold adjustment or resampling technique was applied to correct the imbalance.

For the retail dataset, these issues were not as serious because the class split was closer to 75/25 and the RFM features provided clearer and more separable patterns.

## How It Was Improved

Although Gaussian Naive Bayes has very few tuneable parameters, several steps were taken to make the model as reliable as possible:

- Feature scaling using StandardScaler was applied to Time and Amount in the credit card dataset, since V1 to V28 were already PCA-transformed and on a similar scale,
- Log transformations were applied to Recency, Frequency, and Monetary value in the retail dataset to reduce skewness and help the Gaussian distribution assumption fit better,
- A stratified train-test split was used for both datasets to preserve class proportions in training and testing sets,
- ROC-AUC was added alongside Accuracy, Precision, Recall, and F1-score to give a more complete evaluation, especially for the imbalanced credit card dataset,
- A train vs test performance comparison was included to confirm the model was not overfitting,
- Confusion matrix analysis was included to understand the breakdown of correct and incorrect predictions for both classes.

No threshold adjustment or resampling technique was applied in this implementation. These would be the logical next steps to improve performance on the credit card dataset, where the default decision boundary caused too many false positives.

## Limitations

The Gaussian Naive Bayes model had several limitations:

- It assumes all features are independent, which is not true for RFM features or the PCA-transformed fraud features,
- It assumes features follow a Gaussian distribution, which may not perfectly fit all features even after log transformation,
- It does not handle class imbalance, making it unreliable for fraud detection without additional techniques,
- It has very few hyperparameters to tune, limiting opportunities for improvement,
- High Accuracy can be misleading for imbalanced datasets and should always be evaluated alongside Precision, Recall, and F1-score.

The limitations were much more visible on the Credit Card Fraud dataset than on the Online Retail dataset.

## Final Conclusion

Overall, the Gaussian Naive Bayes model performed well on the Online Retail dataset and produced a mixed result on the Credit Card Fraud dataset.

For the Online Retail dataset, the model achieved strong and balanced results with an F1-score of 0.928 and ROC-AUC of 0.992. This shows that Gaussian Naive Bayes was an effective probabilistic baseline for customer value classification, and that RFM features provided enough signal for the model to work well despite its independence assumption.

For the Credit Card Fraud dataset, the model achieved very high ROC-AUC of 0.989 and strong Recall of 0.933, but Precision of 0.173 and F1-score of 0.292 showed that the model struggled with the imbalanced decision boundary. The model correctly ranked fraud transactions by probability, but the default threshold caused too many false positives.

The main learning from this model was that Gaussian Naive Bayes is a useful and fast probabilistic baseline, but its effectiveness depends heavily on class balance and the degree to which the feature independence assumption holds. For fraud detection, techniques such as threshold adjustment, oversampling, or undersampling would be needed to make the model practical.
